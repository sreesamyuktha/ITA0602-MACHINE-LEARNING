```python

# Candidate-Elimination Algorithm

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

n = len(data[0]) - 1

S = ['0'] * n
G = [['?'] * n]

for row in data:
    x, label = row[:-1], row[-1]

    if label == 'Yes':
        G = [g for g in G if all(
            g[i] == '?' or g[i] == x[i] for i in range(n)
        )]

        for i in range(n):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

    else:
        new_G = []
        for g in G:
            for i in range(n):
                if g[i] == '?':
                    if S[i] != '?' and S[i] != '0' and S[i] != x[i]:
                        h = g.copy()
                        h[i] = S[i]
                        new_G.append(h)
        G = new_G

print("Final Specific Boundary (S):")
print(S)

print("\nFinal General Boundary (G):")
for g in G:
    print(g)
```
